```python
import requests

def get_repo_forks(owner, repo, token=None):
    """
    查詢某 GitHub repo 被哪些帳號 fork

    :param owner: repo 擁有者帳號名稱
    :param repo: repo 名稱
    :param token: GitHub Personal Access Token（建議提供，避免 API 限制）
    :return: fork 過該 repo 的帳號名稱列表
    """
    forks = []
    page = 1
    headers = {"Accept": "application/vnd.github+json"}
    if token:
        headers["Authorization"] = f"token {token}"

    while True:
        url = f"https://api.github.com/repos/{owner}/{repo}/forks?page={page}&per_page=100"
        response = requests.get(url, headers=headers)

        if response.status_code != 200:
            raise Exception(f"GitHub API Error: {response.status_code} - {response.text}")

        data = response.json()
        if not data:
            break

        forks.extend([fork["owner"]["login"] for fork in data])
        page += 1

    return forks
```

```python
# 請用你的 GitHub token，如果沒有也可以不用，只是會比較快 hit rate limit
token = "your_github_token"
owner = "octocat"
repo = "Hello-World"

forkers = get_repo_forks(owner, repo, token)
print(f"Forked by {len(forkers)} users:")
print(forkers)
```
